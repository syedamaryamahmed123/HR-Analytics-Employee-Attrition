```python
attrition_rate = (
    df["Attrition"].eq("Yes").mean() * 100
)

print(f"Overall Attrition Rate: {attrition_rate:.2f}%")
```

```python
print("Dataset Shape:", df.shape)
print("Duplicate Rows:", df.duplicated().sum())
print("Missing Values:", df.isnull().sum().sum())
print("Unique Employees:", df["EmployeeNumber"].nunique())
print("Employees Who Left:", (df["Attrition"] == "Yes").sum())
print("Employees Who Stayed:", (df["Attrition"] == "No").sum())
print(f"Attrition Rate: {attrition_rate:.2f}%")
print("Departments:", df["Department"].nunique())
print("Job Roles:", df["JobRole"].nunique())
```
